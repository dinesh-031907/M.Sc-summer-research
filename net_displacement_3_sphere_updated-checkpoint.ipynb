```python
import numpy as np
```

```python
def oseen_n_force_calculator(state , action , number_of_spheres = 3):
    
    # state : {x1 , x2 , x3}
    # action : {del l1 , del l2}
    
    r = 1 # radius = 1
    u = 1 # coefficient of viscosity = 1
    
    n = number_of_spheres
    H = np.zeros((n,n),dtype = np.float64)
    
    # oseen_matrix_calculator
    for i in range(n):
        for j in range(n):
            if i == j:
                H[i,j] = 1/(6 * r * np.pi * u)
            else:
                H[i,j] = 1/(4 * np.pi * u * abs(state[i] - state[j]))

    # force_calculator 
    b = np.zeros((n))
    A = np.zeros((n,n),dtype = np.float64)

    A[0,:] = H[1,:] - H[0,:]
    A[1,:] = H[2,:] - H[1,:]
    A[2,:] = 1.0
    b[0] = action[0]
    b[1] = action[1]
    b[2] = 0

    F = np.linalg.solve(A,b)

    return H,F
```

```python
def net_disp_finder(l1 , l2 , dl1 , dl2):
    state = np.array([0,l1,l1+l2],dtype = np.float64)
    action = np.array([dl1,dl2],dtype = np.float64)
    H,F = oseen_n_force_calculator(state , action)
    v = H@F
    v_mean = np.mean(v)
    return v_mean , v , F , H
```

```python
cycle = [((6, 6), (4, 0)),
         ((10, 6), (0, 4)),
         ((10, 10), (-4, 0)),
         ((6, 10), (0, -4))]
```

# I have adjusted with the condition that first find small expansion dl = +4 / 1000001 and do this for 1000001 times to get net expansion of +4 units
Here i  used this small dl to update Oseen Tensor and get velovity then use this to find the next position and update the current position .
similarly do this for all steps and here I have got +0.105 units displacement that is different from the previous result that i got without this 

```python
net_disp = 0.0
p = 2
for (l1,l2) , (dl1,dl2) in cycle:
    disp = 0
    if dl2 == 0:
        a = np.linspace(0,dl1,p)
        l1n = l1
        l2n = l2
        for i in a:
            vm , v , F , H = net_disp_finder(l1n ,l2n ,dl1/(p-1) ,dl2/(p-1))
            l1n = l1 + i
            disp += vm
        net_disp += disp
    else:
        a = np.linspace(0,dl2,p)
        l1n = l1
        l2n = l2
        for i in a:
            vm , v , F , H = net_disp_finder(l1n ,l2n ,dl1/(p-1) ,dl2/(p-1))
            l2n = l2 + i
            disp += vm
        net_disp += disp
    print("Current_State")
    print("----",l1 , l2)
    print("-------------",dl1 , dl2)
    print("    .........    ")
    print("displacement =" ,v[1])
```
