```python
import numpy as np
```

```python
def oseen_n_force_calculator(state , action , number_of_spheres):
    
    # state : {x1 , x2 , x3}
    # action : {del l1 , del l2}
    
    r = 1 # radius = 1
    u = 1 # coefficient of viscosity = 1
    
    n = number_of_spheres
    H = np.zeros((n,n),dtype = np.float64)
    
    # oseen_matrix_calculator
    for i in range(n):
        for j in range(n):
            if i == j:
                H[i,j] = 1/(6 * r * np.pi * u)
            else:
                H[i,j] = 1/(4 * np.pi * u * abs(state[i] - state[j]))

    # force_calculator 
    b = np.zeros((n))
    A = np.zeros((n,n),dtype = np.float64)

    for i in range(n-1):
        A[i,:] = H[i+1,:] - H[i,:]
        b[i] = action[i]
    
    A[n-1,:] = 1
    b[n-1] = 0

    F = np.linalg.solve(A,b)

    return H,F
```

```python
def net_disp_finder(l1 , l2 , dl1 , dl2):
    state = np.array([0,l1,l1+l2],dtype = np.float64)
    action = np.array([dl1,dl2],dtype = np.float64)
    H,F = oseen_n_force_calculator(state , action , 3) # 3 sphere case
    v = H@F
    v_mean = np.mean(v)
    return v_mean , v , F , H
```

```python
cycle = [((10, 10), (-4, 0)),
         ((6, 10), (0, -4)),
         ((6, 6), (4, 0)),
         ((10, 6), (0, 4))]
```

```python
net_disp = 0.0
for (l1,l2) , (dl1,dl2) in cycle:
    disp , vel , F , H = net_disp_finder(l1,l2,dl1,dl2)
    net_disp += vel[1]
    print("Position of middle sphere= ",vel[1])
    print("Current_State-------",l1 , "          ", l2)
    print("Action -------------",dl1 ,"          ", dl2)
    print("    .........   ")
```
