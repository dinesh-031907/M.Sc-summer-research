# For Finding Rewards : 
    Here we consider the rewards as the net displacement of the middle sphere in one step

```python
# - For finding rewards : Here we consider the rewards as the net displacement of the middle sphere in one step/stroke

def oseen_n_force_calculator(state , action , number_of_spheres):
    
    # state : {x1 , x2 , x3}
    # action : {del l1 , del l2}
    
    r = 1 # radius = 1
    u = 1 # coefficient of viscosity = 1
    
    n = number_of_spheres
    H = np.zeros((n,n),dtype = np.float64)
    
    # oseen_matrix_calculator
    for i in range(n):
        for j in range(n):
            if i == j:
                H[i,j] = 1/(6 * r * np.pi * u)
            else:
                H[i,j] = 1/(4 * np.pi * u * abs(state[i] - state[j]))

    # force_calculator 
    b = np.zeros((n))
    A = np.zeros((n,n),dtype = np.float64)

    for i in range(n-1):
        A[i,:] = H[i+1,:] - H[i,:]
        b[i] = action[i]
    
    A[n-1,:] = 1
    b[n-1] = 0

    F = np.linalg.solve(A,b)

    return H,F
```

```python
def net_disp_finder(l1 , l2 , dl1 , dl2):
    state = np.array([0,l1,l1+l2],dtype = np.float64)
    action = np.array([dl1,dl2],dtype = np.float64)
    H,F = oseen_n_force_calculator(state , action , 3) # 3 sphere case
    v = H@F
    v_mean = np.mean(v)
    return v_mean , v , F , H
```

```python
cycle = [((10, 10), (-4, 0)),
         ((6, 10), (0, -4)),
         ((6, 6), (4, 0)),
         ((10, 6), (0, 4))]
```

```python
net_disp = 0.0
for (l1,l2) , (dl1,dl2) in cycle:
    disp , vel , F , H = net_disp_finder(l1,l2,dl1,dl2)
    net_disp += disp
    print("Change in position of middle sphere = ",np.round(vel[1],4),"units")
    print("Change in position of Centroid      = ",np.round(disp,4),"units")
    print("Current_State-------",l1 , "          ", l2)
    print("Action -------------",dl1 ,"          ", dl2)
    print("    .........   ")
    print("    .........   ")
    print("    .........   ")
```

```python
# transition = { current_state : {action1 : next_state , ... } ... }
transition = {
    1 : { 1 : None , 
          2 : 2 ,
          3 : None ,
          4 : 4
        },
    2 : { 1 : 1 ,
          2 : None ,
          3 : None ,
          4 : 3 
        },
    3 : { 1 : 4 ,
          2 : None ,
          3 : 2 ,
          4 : None
        },
    4 : { 1 : None ,
          2 : 3 ,
          3 : 1 ,
          4 : None 
        }
}
# rewards = { current_state : {action : reward , ... } ... }
rewards = {
    1 : {
        2 : -1.35 ,
        4 : 1.35
    },
    2 : {
        1 : 1.35 ,
        4 : 1.44 
    },
    3 : {
        1 : 1.44 ,
        3 : -1.44
    },
    4 : {
        2 : -1.44 ,
        3 : -1.35
    }
} 
```

# INITIALIZING Q-MATRIX

```python
# q_matrix must have some elements be zero because those actions are prohibited in that particular state in concern :
zero_position = [(0,0) , (0,2) ,(1,1) ,(1,2) , (2,1) , (2,3) , (3,0) , (3,3)]
for pos in zero_position:
    q_matrix[pos] = 0
```

```python
# Hyper_parameters:
```

```python
alpha = 1  # ideal learning factor for most of rl q-network  
gamma = 0.99   # To increase the importance of future reward than immediate reward
epsilon = 0.2  # for e-greedy method
episodes = 2000  
epsilon_rate = 0.999  # so that when a good policy is learned in the future the swimmer tends mostly towards that policy.
```

```python
# Datas for plotting :
cumulative_disp = 0
disp_history = []
i = 0
```

```python
# GIVEN CURRENT Q-MATRIX IT FINDS OUT , WHAT POLICY CURRENTLY DOES BEST OPTIMAL ACTIONS TO GET MAXIMUM REWARD IN THE CURRENT STATE :
# HERE REWARD IS DISPLACEMENT PER STEP

def trace_policy_until_cycle(Q, transition, rewards, start_state, max_s=10):
    state = start_state + 1 
    seen = []
    tot_d = 0

    for _ in range(max_s):

        seen.append(state)

        q_values = Q[state - 1, :]  
        valid_actions = [a for a in transition[state] if transition[state][a] is not None]

        best_action = max(valid_actions, key=lambda a: q_values[a - 1])
        tot_d += rewards[state].get(best_action, 0)

        state = transition[state][best_action]

    tot_d /= len(seen)
    return tot_d 
```

```python
for episode in range(episodes):
    i += 1
    state = np.random.randint(0,num_states-1) if len(disp_history) == 0 else next_state
    state_1 = state + 1
    valid_actions = [a for a in range(num_actions) if transition[state_1][a+1] is not None]
    if random.random() < epsilon:
        action = random.choice(valid_actions)
    else:
        action = valid_actions[np.argmax([q_matrix[state, a] for a in valid_actions])]
    action_1 = action + 1
    next_state_1 = transition[state_1][action_1]
    next_state = next_state_1 - 1
    reward = rewards[state_1][action_1]
    next_valid = [a for a in range(num_actions)
                      if transition[next_state_1][a + 1]]
    max_q_next = max([q_matrix[next_state, a] for a in next_valid])
    q_matrix[state, action] += alpha * (reward + gamma * max_q_next - q_matrix[state, action])
    D = trace_policy_until_cycle(q_matrix,transition,rewards,state)
    cumulative_disp += D
    disp_history.append(cumulative_disp)
    epsilon *= epsilon_rate
```
